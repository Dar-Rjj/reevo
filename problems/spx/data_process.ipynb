```python
import os
from os import path

basepath = path.join(os.getcwd(), "export_us")
```

```python
import os
import pandas as pd
import numpy as np

stock_codes = pd.read_csv('sp500_companies.csv')['Symbol']

dfs = []
for stock_code in stock_codes:
    fname = f"{stock_code}.csv"
    fpath = os.path.join(basepath, fname)
    if os.path.exists(fpath):
        df = pd.read_csv(fpath, parse_dates=['日期'])[['日期','开盘','最高','最低','收盘','成交量','成交额']]
        df.columns = ['date', 'open', 'high', 'low', 'close', 'volume', 'amount']
        df = df[(df['date'] >= '2015-06-30') & (df['date'] <= '2020-01-31')]
        df['stock_code'] = stock_code
        dfs.append(df)

train_df = pd.concat(dfs, ignore_index=True)
train_df.set_index(['stock_code', 'date'], inplace=True)
train_df.sort_index(inplace=True)

train_df.to_csv("train_data.csv")
print(train_df)
```

```python
import os
import pandas as pd
import numpy as np

stock_codes = pd.read_csv('sp500_companies.csv')['Symbol']

dfs = []
for stock_code in stock_codes:
    fname = f"{stock_code}.csv"
    fpath = os.path.join(basepath, fname)
    if os.path.exists(fpath):
        df = pd.read_csv(fpath, parse_dates=['日期'])[['日期','开盘','最高','最低','收盘','成交量','成交额']]
        df.columns = ['date', 'open', 'high', 'low', 'close', 'volume', 'amount']
        df = df[(df['date'] >= '2020-06-30') & (df['date'] <= '2024-01-31')]
        df['stock_code'] = stock_code
        dfs.append(df)

test_df = pd.concat(dfs, ignore_index=True)
test_df.set_index(['stock_code', 'date'], inplace=True)
test_df.sort_index(inplace=True)

test_df.to_csv("test_data.csv")
print(test_df)
```
